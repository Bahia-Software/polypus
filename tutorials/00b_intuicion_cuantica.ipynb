## Bit clásico frente a qubit

Un bit clásico, el de un ordenador convencional, vale **0 o 1**, siempre uno de los dos, de forma determinista: al leerlo, se obtiene siempre el mismo valor mientras no cambie.

Un **qubit** puede estar en una combinación de 0 y 1 al mismo tiempo. A esto se le llama **superposición**. No es que "no se sepa" si vale 0 o 1: mientras está en superposición, genuinamente no tiene un valor definido todavía.

## Superposición

Una analogía útil, aunque imperfecta, es la de una moneda **girando en el aire**, todavía sin caer. Mientras la moneda gira, no es ni cara ni cruz: está en un estado intermedio genuino. Al caer y observarla, se ve forzada a un resultado concreto, cara o cruz.

Un qubit en superposición es algo parecido: un estado genuinamente intermedio entre 0 y 1, que solo se convierte en un valor concreto **al medirlo**.

La puerta que pone un qubit en superposición se llama **H**, de *Hadamard*. A continuación, un ejemplo real con Polypus:

```python
import polypus

qc = polypus.Circuit(1).h(0).measure_all()
result = polypus.run_quantum_circuit(qc, shots=1000, infrastructure="local")
print(result.counts[0])  # el [0] coge el primer resultado, el único aquí; se explica en el próximo notebook
```

El resultado se parece a `{'0': ~500, '1': ~500}`, aunque no exactamente esos números: cada ejecución de la celda da valores ligeramente distintos, porque es aleatoriedad real, no un resultado fijo.

Lo que acaba de pasar: `h(0)` puso el qubit 0 en superposición, y `shots=1000` significa "mide el resultado 1000 veces". Cada una de esas 1000 medidas fuerza al qubit a caer en 0 o en 1, más o menos al 50%, y `result.counts` da cuántas veces salió cada uno. Por eso el resultado es un diccionario de frecuencias, no un único valor.

## Medida y colapso

Al hecho de que medir un qubit lo fuerce a un valor concreto se le llama **colapso**: antes de medir, el qubit está en superposición; al medir, colapsa a 0 o a 1, y a partir de ahí ya es un valor normal y corriente, como un bit clásico.

Por eso en cuántica casi nunca se ejecuta un circuito una sola vez: se ejecuta muchas veces, los `shots`, para ver la **distribución de probabilidad** del resultado, tal como se ha visto arriba.

## Puertas: H y CX/CNOT

Una puerta cuántica transforma el estado de uno o varios qubits. Las dos necesarias para el próximo notebook:

- **H**: pone un qubit en superposición, lo visto en el apartado anterior.
- **CX**, también llamada CNOT: actúa sobre dos qubits, uno de "control" y uno "objetivo". Si el control vale 1, invierte el objetivo; si vale 0, lo deja igual. Es una puerta *condicional*.

La parte interesante ocurre cuando el qubit de control está en superposición, tras un H: entonces CX no puede "decidir" de forma clásica si invertir o no el objetivo, y el resultado es que los dos qubits quedan **entrelazados**.

## Entrelazamiento

El entrelazamiento es la correlación entre dos qubits por la que el resultado de medir uno indica algo del otro, aunque cada uno por separado siga pareciendo aleatorio.

Este ejemplo no se ejecuta todavía aquí: es justo el resultado sorpresa del ejemplo central del próximo notebook, donde dos qubits, tras un H y un CX, **solo** dan como resultado `00` o `11`, nunca `01` ni `10`. Con la intuición de arriba ya queda explicado lo necesario para entenderlo llegado el momento.

## Notación de Dirac

Esta notación aparece en el resto de la serie, así que conviene verla una vez aunque no sea necesaria todavía para el próximo notebook:

- `|0⟩` y `|1⟩` son las dos etiquetas para "el qubit vale 0" y "el qubit vale 1". Los símbolos `|` y `⟩` son solo una forma estándar de escribir "esto es un estado cuántico", nada más.
- `|+⟩` es la etiqueta habitual para el estado de superposición que crea la puerta H, el estado "entre 0 y 1" visto arriba.
