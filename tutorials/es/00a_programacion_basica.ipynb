## Variables


Una variable guarda un valor con un nombre, para poder usarlo después.

````python
numero_de_qubits = 2
nombre_backend = "local"

print(numero_de_qubits)
print(nombre_backend)
````

El valor de `nombre_backend` es texto, y el texto, un **string**, va siempre entre comillas. `"local"` es un string; `local` sin comillas sería, si existiese, el nombre de una variable, y Python daría error si no existiera.

## Funciones y argumentos posicionales

Una función es un bloque de código con nombre al que se le pueden pasar valores, los **argumentos**, entre paréntesis. Cuando los argumentos van en un orden fijo, sin nombre, se llaman **argumentos posicionales**.

````python
def sumar(a, b):
    return a + b


print(sumar(3, 4))
````

Un detalle importante para lo que viene: en programación, casi todo se cuenta **empezando desde 0**, no desde 1. El primer elemento es el elemento número 0, el segundo es el número 1, y así sucesivamente. Más adelante, expresiones como `h(0)`, para referirse al primer qubit de un circuito, siguen exactamente esta misma convención.

## Objetos, métodos, y encadenar llamadas

Un objeto agrupa datos y comportamiento bajo un mismo nombre: guarda valores propios y ofrece funciones propias, los **métodos**, a las que se accede con un punto, `objeto.metodo()`. El `Robot` de abajo es un ejemplo: guarda una lista de pasos dados, y sus métodos son las acciones para moverse.

Muchos métodos, además de hacer algo, **devuelven el propio objeto**, lo que permite encadenar varias llamadas seguidas, una detrás de otra.

````python
class Robot:
    def __init__(self):
        self.pasos = []

    def mover(self):
        self.pasos.append("adelante")
        return self  # devuelve el propio robot, para poder seguir encadenando

    def girar(self):
        self.pasos.append("girar")
        return self


r = Robot()
r.mover().girar().mover()  # tres llamadas encadenadas, una detrás de otra
print(r.pasos)
````

Este truco, `return self`, es exactamente el que usa `polypus.Circuit`: cada método que añade una puerta devuelve el propio circuito, así que es posible escribir `polypus.Circuit(2).h(0).cx(0, 1).measure_all()` en una sola línea, con el mismo mecanismo que se acaba de ver con el robot.

## Argumentos con nombre

Al definir una función, un parámetro puede tener un **valor por defecto**, como `saludo="Hola"` en el ejemplo siguiente: si ese argumento no se indica al llamar a la función, se usa el valor por defecto.

Además, en lugar de depender del orden, los argumentos se pueden pasar por su nombre, con `nombre=valor`. A esto se le llama **argumento con nombre**, o *keyword argument*. Es especialmente útil cuando una función tiene varios argumentos y conviene dejar claro cuál es cuál, o cuando solo se quiere fijar uno de varios argumentos opcionales sin tocar los demás.

````python
def saludar(nombre, saludo="Hola"):
    print(f"{saludo}, {nombre}!")


saludar("Ana")
saludar("Ana", saludo="Buenas tardes")
````

Más adelante aparecerán llamadas como `run_quantum_circuit(qc, shots=1000, infrastructure="local")`. Ahí, `shots` e `infrastructure` son argumentos con nombre: queda claro qué valor es cuál, sin depender de memorizar en qué posición va cada uno.

## Atributos frente a métodos

Un objeto también puede tener **atributos**: valores que guarda, a los que se accede con un punto pero **sin paréntesis**. Es fácil confundirlos con los métodos, así que conviene verlos uno al lado del otro.

````python
class Paquete:
    def __init__(self, peso):
        self.peso = peso  # atributo: se lee sin paréntesis

    def describir(self):  # método: se llama con paréntesis
        return f"Paquete de {self.peso} kg"


p = Paquete(5)
print(p.peso)  # atributo
print(p.describir())  # método
````

Más adelante, `result.counts`, sin paréntesis porque es un atributo, se usa para leer el resultado de ejecutar un circuito, no `result.counts()`.

## Diccionarios

Un diccionario guarda pares de **clave → valor**. Se accede a un valor por su clave, entre corchetes.

````python
frutas = {"manzanas": 3, "peras": 5}

print(frutas["manzanas"])

for fruta, cantidad in frutas.items():
    print(fruta, "->", cantidad)
````

Esto es exactamente lo que aparece al ejecutar un circuito cuántico: el resultado es un diccionario que asocia cada resultado medido, un string como `"00"`, con el número de veces que salió.
