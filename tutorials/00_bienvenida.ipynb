# Bienvenido/a a los tutoriales de Polypus

Polypus es una librería para construir y ejecutar circuitos cuánticos de manera distribuída: el mismo código sirve para correr en un simulador en tu propio ordenador, repartir el trabajo entre varios simuladores a la vez, o ejecutar en un ordenador cuántico real, sin cambiar nada del circuito.

Esta serie de tutoriales componen una guía sobre como usar Polypus **desde cero**, tanto si vienes del mundo de la física como de la programación, o de ninguna de las dos. No hace falta que sepas programar ni que hayas estudiado mecánica cuántica en profundidad antes, está pensada para que cualquiera pueda seguirla de manera íntegra. Si ya tienes experiencia en alguna de las dos áreas, podrás saltarte las partes que no necesites.

## Cómo funciona esta serie

Toda la serie son notebooks como este, enlazados entre sí: cada uno termina con un enlace al siguiente, así que se puede avanzar entre ellos simplemente haciendo click, sin tener que gestionar los archivos por tu cuenta.

Estos notebooks viven en la carpeta [`tutorials/`](.) del repositorio de Polypus en GitHub. Para seguirlos en tu propio Jupyter, JupyterLab o VS Code, descarga esta carpeta — clonando el repositorio (ver el comando en el [README principal](../README.md)) o con "Download ZIP" desde GitHub. No necesitas compilar nada del código Rust para esto, solo tener la carpeta en tu ordenador.

Más adelante también podrás abrirlos directamente en Google Colab sin descargar nada.

## Antes de empezar: instalación

Instala Polypus con pip:

```bash
pip install polypus-quantum
```

(El nombre del paquete en PyPI es `polypus-quantum`; dentro de Python se importa como `import polypus`.)

## Elige tu punto de partida

Antes de arrancar, es importante prestar atención a estas dos preguntas:

1. ¿Has programado antes, aunque sea un poco (en Python o en cualquier otro lenguaje)?
2. ¿Conoces ya conceptos básicos de mecánica cuántica (qubit, superposición, medida)?

Según tus respuestas, tu punto de partida con Polypus es distinto:

| ¿Programación? | ¿Cuántica? | Por dónde empezar |
|---|---|---|
| No | No | [`00a_programacion_basica.ipynb`](00a_programacion_basica.ipynb) → [`00b_intuicion_cuantica.ipynb`](00b_intuicion_cuantica.ipynb) → [`01_primer_circuito.ipynb`](01_primer_circuito.ipynb) |
| No | Sí | [`00a_programacion_basica.ipynb`](00a_programacion_basica.ipynb) → [`01_primer_circuito.ipynb`](01_primer_circuito.ipynb) |
| Sí | No | [`00b_intuicion_cuantica.ipynb`](00b_intuicion_cuantica.ipynb) → [`01_primer_circuito.ipynb`](01_primer_circuito.ipynb) |
| Sí | Sí | directamente [`01_primer_circuito.ipynb`](01_primer_circuito.ipynb) |

**Si te faltan las dos bases, empieza siempre por `00a` antes que por `00b`.** `00b` da por hecho que ya sabes ejecutar una celda y llamar a una función — eso es justo lo que enseña `00a`.
