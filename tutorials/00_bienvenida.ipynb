# Bienvenido/a a los tutoriales de Polypus

Polypus es una librería para construir y ejecutar circuitos cuánticos de manera distribuida: el mismo código sirve para correr en un simulador en un ordenador local, repartir el trabajo entre varios simuladores a la vez, o ejecutar en un ordenador cuántico real, sin cambiar nada del circuito.

Esta serie de tutoriales es una guía sobre cómo usar Polypus **desde cero**, tanto para quien llega desde la física como para quien llega desde la programación, o para quien no llega desde ninguna de las dos. No hace falta saber programar ni haber estudiado mecánica cuántica en profundidad de antemano: está pensada para que cualquiera pueda seguirla de manera íntegra. Quien ya tenga experiencia en alguna de las dos áreas podrá saltarse las partes que no necesite.

## Cómo funciona esta serie

Toda la serie son notebooks como este, enlazados entre sí: cada uno termina con un enlace al siguiente, así que se puede avanzar entre ellos simplemente haciendo click, sin necesidad de gestionar los archivos por separado.

Este contenido se mantiene sincronizado automáticamente con cada versión publicada de Polypus, así que siempre refleja la API actual. Para seguir los notebooks en un Jupyter, JupyterLab o VS Code propios, basta con descargar este mismo repositorio, clonándolo o con "Download ZIP" desde GitHub: el contenido de esta carpeta son solo notebooks, no hace falta instalar ni compilar nada relacionado con Rust.

Más adelante también será posible abrirlos directamente en Google Colab sin descargar nada.

## Antes de empezar: instalación

Instalación de Polypus con pip:

```bash
pip install polypus-quantum
```

El nombre del paquete en PyPI es `polypus-quantum`; dentro de Python se importa como `import polypus`.

## Elegir el punto de partida

Antes de arrancar, conviene tener en cuenta dos cosas:

1. Experiencia previa programando, aunque sea mínima, en Python o en cualquier otro lenguaje.
2. Conocimiento de conceptos básicos de mecánica cuántica: qubit, superposición, medida.

Según el caso, el punto de partida con Polypus es distinto:

| ¿Programación? | ¿Cuántica? | Por dónde empezar |
|---|---|---|
| No | No | [`00a_programacion_basica.ipynb`](00a_programacion_basica.ipynb) → [`00b_intuicion_cuantica.ipynb`](00b_intuicion_cuantica.ipynb) → [`01_primer_circuito.ipynb`](01_primer_circuito.ipynb) |

**Si faltan las dos bases, conviene empezar siempre por [`00a`](00a_programacion_basica.ipynb) antes que por [`00b`](00b_intuicion_cuantica.ipynb).** [`00b`](00b_intuicion_cuantica.ipynb) da por hecho que ya es posible ejecutar una celda y llamar a una función: eso es justo lo que enseña [`00a`](00a_programacion_basica.ipynb).
