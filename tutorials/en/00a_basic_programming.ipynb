## Functions and positional arguments

A function is a named block of code that can be passed values, the **arguments**, inside parentheses. When the arguments go in a fixed order, without a name, they are called **positional arguments**.

````python
def add(a, b):
    return a + b


print(add(3, 4))
````

An important detail for what comes next: in programming, almost everything is counted **starting from 0**, not from 1. The first element is element number 0, the second is number 1, and so on. Later on, expressions like `h(0)`, referring to the first qubit of a circuit, follow this exact same convention.

## Objects, methods, and chaining calls

An object groups data and behavior under a single name: it stores its own values and offers its own functions, the **methods**, accessed with a dot, `object.method()`. The `Robot` below is an example: it stores a list of steps taken, and its methods are the actions to move around.

Many methods, besides doing something, **return the object itself**, which makes it possible to chain several calls one after another.

````python
class Robot:
    def __init__(self):
        self.steps = []

    def move(self):
        self.steps.append("forward")
        return self  # returns the robot itself, to keep chaining

    def turn(self):
        self.steps.append("turn")
        return self


r = Robot()
r.move().turn().move()  # three calls chained one after another
print(r.steps)
````

This trick, `return self`, is exactly what `polypus.Circuit` uses: every method that adds a gate returns the circuit itself, so it is possible to write `polypus.Circuit(2).h(0).cx(0, 1).measure_all()` in a single line, with the same mechanism just shown with the robot.

## Keyword arguments

When defining a function, a parameter can have a **default value**, like `greeting="Hello"` in the example below: if that argument is not given when calling the function, the default value is used.

Also, instead of relying on order, arguments can be passed by name, with `name=value`. This is called a **keyword argument**. It is especially useful when a function has several arguments and it helps to make clear which is which, or when only one of several optional arguments needs to be set without touching the rest.

````python
def greet(name, greeting="Hello"):
    print(f"{greeting}, {name}!")


greet("Ana")
greet("Ana", greeting="Good afternoon")
````

Later on, calls like `run_quantum_circuit(qc, shots=1000, infrastructure="local")` will appear. There, `shots` and `infrastructure` are keyword arguments: it is clear which value is which, without having to memorize the position of each one.

## Attributes versus methods

An object can also have **attributes**: values it stores, accessed with a dot but **without parentheses**. It is easy to confuse them with methods, so it helps to see them side by side.

````python
class Package:
    def __init__(self, weight):
        self.weight = weight  # attribute: read without parentheses

    def describe(self):  # method: called with parentheses
        return f"Package of {self.weight} kg"


p = Package(5)
print(p.weight)  # attribute
print(p.describe())  # method
````

Later on, `result.counts`, without parentheses because it is an attribute, is used to read the result of running a circuit, not `result.counts()`.

## Dictionaries

A dictionary stores **key → value** pairs. A value is accessed by its key, in square brackets.

````python
fruits = {"apples": 3, "pears": 5}

print(fruits["apples"])

for fruit, count in fruits.items():
    print(fruit, "->", count)
````

This is exactly what shows up when running a quantum circuit: the result is a dictionary that maps each measured outcome, a string like `"00"`, to the number of times it came up.
