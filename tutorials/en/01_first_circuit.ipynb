# First quantum circuit

This notebook builds, runs, and explains a real quantum circuit with Polypus: the Bell state, the most cited example in all of quantum computing.

## Building the circuit

A Polypus circuit is built by chaining methods on `Circuit`, as seen in [`00a`](00a_basic_programming.ipynb):

- `Circuit(2)`: a circuit with 2 qubits, numbered 0 and 1.
- `.h(0)`: puts qubit 0 in superposition.
- `.cx(0, 1)`: applies CX with qubit 0 as control and qubit 1 as target.
- `.measure_all()`: measures all qubits at the end.

```python
import polypus

bell = polypus.Circuit(2).h(0).cx(0, 1).measure_all()
```

This circuit, H followed by CX on a pair of qubits initialized in `|00⟩`, is the so-called **Bell state**, or **Bell pair**: the minimal construction that produces entanglement, and it comes back later in the series, for example in [`04a_teleportation.ipynb`](04a_teleportation.ipynb).

## Running the circuit

`run_quantum_circuit` runs a circuit and returns its result. It takes the circuit as its first positional argument, `bell` in this case, plus two main keyword arguments:

- `shots`: how many times the circuit is run and measured. With more shots, the frequency distribution gets closer to the real probability, as seen in [`00b`](00b_quantum_intuition.ipynb).
- `infrastructure`: where it runs. `"local"` uses a simulator on the computer itself; the other options are explained in [`03a_local_execution.ipynb`](03a_local_execution.ipynb).

```python
result = polypus.run_quantum_circuit(bell, shots=1000, infrastructure="local")
print(result.counts)
```

The result looks like `[{'11': ~500, '00': ~500}]`, though not exactly those numbers: it is real randomness. `result.counts` comes wrapped in a **list** with a single element. Why, and what changes when the run is spread across several QPUs, is explained in [`03a_local_execution.ipynb`](03a_local_execution.ipynb).

To get just the dictionary, access position 0:

```python
counts = result.counts[0]
print(counts)
```

## The surprise result

Only two keys show up in `counts`, `'00'` and `'11'`, never `'01'` or `'10'`. Each qubit on its own still looks random, with roughly half the results landing on 0 and half on 1, but **the two qubits always agree**: this is entanglement, as anticipated in [`00b`](00b_quantum_intuition.ipynb). Measuring one of the two qubits instantly determines the result of the other.

## Visualizing the result

A histogram helps see the distribution at a glance. There is no need to know matplotlib in depth for this, just this recipe:

```python
import matplotlib.pyplot as plt

plt.bar(counts.keys(), counts.values())
plt.xlabel("Measured result")
plt.ylabel("Number of times")
plt.title("Result distribution of the Bell state")
plt.show()
```

## Summary

This notebook built a circuit by chaining methods, ran it with `run_quantum_circuit`, and read and interpreted its result: two qubits that, after an H gate and a CX, end up entangled, a phenomenon with no classical equivalent.
