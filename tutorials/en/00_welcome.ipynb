# Welcome to the Polypus tutorials

Polypus is a library for building and running quantum circuits in a distributed way: the same code runs on a simulator on a local computer, spreads the work across several simulators at once, or runs on a real quantum computer, without changing anything in the circuit.

This series of tutorials is a guide to using Polypus **from scratch**, for readers coming from physics, from programming, or from neither. No prior programming experience or deep knowledge of quantum mechanics is required beforehand: it is designed so that anyone can follow it in full. Readers who already have experience in either area can skip the parts they do not need.

*Este notebook también está disponible en español: [`00_bienvenida.ipynb`](../es/00_bienvenida.ipynb).*

## How this series works

The whole series consists of notebooks like this one, linked to each other: each one ends with a link to the next, so it is possible to move through them just by clicking, without having to manage the files separately.

This content stays automatically in sync with every published release of Polypus, so it always reflects the current API. To follow the notebooks in a local Jupyter, JupyterLab, or VS Code, just download this same repository, either by cloning it or with "Download ZIP" from GitHub: the content of this folder is notebooks only, no Rust installation or compilation needed.

Later on it will also be possible to open them directly in Google Colab without downloading anything.

## Before starting: installation

Install Polypus with pip:

```bash
pip install polypus-quantum
```

The package name on PyPI is `polypus-quantum`; inside Python it is imported as `import polypus`.

## Choosing a starting point

Before starting, it helps to know two things:

1. Prior programming experience, even minimal, in Python or in any other language.
2. Knowledge of basic quantum mechanics concepts: qubit, superposition, measurement.

Depending on the case, the starting point with Polypus is different:

| Programming? | Quantum? | Where to start |
|---|---|---|
| No | No | [`00a_basic_programming.ipynb`](00a_basic_programming.ipynb) → [`00b_quantum_intuition.ipynb`](00b_quantum_intuition.ipynb) → [`01_first_circuit.ipynb`](01_first_circuit.ipynb) |
| No | Yes | [`00a_basic_programming.ipynb`](00a_basic_programming.ipynb) → [`01_first_circuit.ipynb`](01_first_circuit.ipynb) |
| Yes | No | [`00b_quantum_intuition.ipynb`](00b_quantum_intuition.ipynb) → [`01_first_circuit.ipynb`](01_first_circuit.ipynb) |
| Yes | Yes | straight to [`01_first_circuit.ipynb`](01_first_circuit.ipynb) |

**If both are missing, it is best to always start with [`00a`](00a_basic_programming.ipynb) before [`00b`](00b_quantum_intuition.ipynb).** [`00b`](00b_quantum_intuition.ipynb) assumes it is already possible to run a cell and call a function: that is exactly what [`00a`](00a_basic_programming.ipynb) teaches.

Readers who already know how to build and run circuits with Qiskit can skip the whole series and go straight to [`from_qiskit_to_polypus.ipynb`](from_qiskit_to_polypus.ipynb), a reference guide with a direct mapping between the two.
