## Classical bit versus qubit

A classical bit, the kind in a conventional computer, is **0 or 1**, always one of the two, deterministically: reading it always gives the same value while it does not change.

A **qubit** can be in a combination of 0 and 1 at the same time. This is called **superposition**. It is not that "it isn't known" whether it is 0 or 1: while in superposition, it genuinely does not have a defined value yet.

## Superposition

A useful, if imperfect, analogy is a coin **spinning in the air**, not yet landed. While the coin spins, it is neither heads nor tails: it is in a genuine in-between state. When it lands and is observed, it is forced into a concrete outcome, heads or tails.

A qubit in superposition is something similar: a genuinely in-between state between 0 and 1, which only turns into a concrete value **upon measurement**.

The gate that puts a qubit in superposition is called **H**, for *Hadamard*. Here is a real example with Polypus:

```python
import polypus

qc = polypus.Circuit(1).h(0).measure_all()
result = polypus.run_quantum_circuit(qc, shots=1000, infrastructure="local")
print(result.counts[0])  # the [0] takes the first result, the only one here; explained in the next notebook
```

The result looks like `{'0': ~500, '1': ~500}`, though not exactly those numbers: every run of the cell gives slightly different values, because it is real randomness, not a fixed result.

What just happened: `h(0)` put qubit 0 in superposition, and `shots=1000` means "measure the result 1000 times." Each of those 1000 measurements forces the qubit to land on 0 or 1, roughly 50/50, and `result.counts` gives how many times each one came up. That is why the result is a frequency dictionary, not a single value.

## Measurement and collapse

The fact that measuring a qubit forces it into a concrete value is called **collapse**: before measuring, the qubit is in superposition; upon measuring, it collapses to 0 or to 1, and from then on it is a value just like any other, like a classical bit.

That is why in quantum computing a circuit is almost never run just once: it is run many times, the `shots`, to see the **probability distribution** of the result, as seen above.

## Gates: H and CX/CNOT

A quantum gate transforms the state of one or more qubits. The two needed for [`01_first_circuit.ipynb`](01_first_circuit.ipynb):

- **H**: puts a qubit in superposition, as seen in the previous section.
- **CX**, also called CNOT: acts on two qubits, one "control" and one "target." If the control is 1, it flips the target; if it is 0, it leaves it unchanged. It is a *conditional* gate.

The interesting part happens when the control qubit is in superposition, after an H: then CX cannot "decide" classically whether to flip the target or not, and the result is that the two qubits end up **entangled**.

## Entanglement

Entanglement is the correlation between two qubits by which the result of measuring one tells something about the other, even though each one on its own still looks random.

This example does not run here yet: it is exactly the surprise result of the central example in [`01_first_circuit.ipynb`](01_first_circuit.ipynb), where two qubits, after an H and a CX, give **only** `00` or `11` as a result, never `01` or `10`. ## Dirac notation

This notation appears throughout the rest of the series, so it is worth seeing once even though it is not needed yet for [`01_first_circuit.ipynb`](01_first_circuit.ipynb):

- `|0⟩` and `|1⟩` are the two labels for "the qubit is 0" and "the qubit is 1." The `|` and `⟩` symbols are just a standard way of writing "this is a quantum state," nothing more.
- `|+⟩` is the usual label for the superposition state created by the H gate, the "between 0 and 1" state seen above.

There is no need to handle the algebra behind it to follow this series: for now, it is enough to take these as names for states, the same way `"local"` is the name of a backend.
